```python
import pandas as pd

#load each dataset
empres_fmd = pd.read_csv("../data/raw_data/empres_fmd.csv")
empres_ppr = pd.read_csv("../data/raw_data/empres_ppr.csv")
empres_lsd = pd.read_csv("../data/raw_data/empres_lsd.csv")
empres_cbpp = pd.read_csv("../data/raw_data/empres_cbpp.csv")
empres_rvf = pd.read_csv("../data/raw_data/empres_rvf.csv", on_bad_lines='skip')
empres_all = pd.read_csv("../data/raw_data/empres_all.csv", on_bad_lines='skip',skiprows=11)
faostat = pd.read_csv("../data/raw_data/faostat_livestock.csv")
land = pd.read_csv("../data/raw_data/faostat_land_area.csv")
climate = pd.read_csv("../data/raw_data/nasa_climate.csv")

#shapes
print("EMPRES-i_fmd: ", empres_fmd.shape)
print("EMPRES-i_ppr: ", empres_ppr.shape)
print("EMPRES-i_lsd: ", empres_lsd.shape)
print("EMPRES-i_cbpp: ", empres_cbpp.shape)
print("EMPRES-i_rvf: ", empres_rvf.shape)
print("EMPRES-i: ", empres_all.shape)
print("FAOSTAT livestock: ", faostat.shape)
print("Land area: ", land.shape)
print("Climate: ", climate.shape)
```
